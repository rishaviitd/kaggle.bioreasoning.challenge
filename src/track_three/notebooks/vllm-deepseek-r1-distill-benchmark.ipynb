```python
######################## PRE FINE-TUNING PARSER #########################

import re

def parse_label(text: str) -> str:
    """
    Extracts up, down, or none from [[ ## label ## ]].
    Returns 'ERROR' if parsing fails.
    """
    if not isinstance(text, str) or not text.strip():
        return "ERROR"

    match = re.search(
        r"\[\[\s*##\s*label\s*##\s*\]\](.*?)(?=\[\[|$)",
        text,
        flags=re.IGNORECASE | re.DOTALL,
    )

    if not match:
        return "ERROR"

    block_content = match.group(1).strip().lower()

    if not re.fullmatch(r"(up|down|none)", block_content):
        return "ERROR"

    return block_content
```

## REJECTION SAMPLING ON BATCH 

```python
import os
import pandas as pd
from vllm import SamplingParams

TRAIN_CSV = "/kaggle/input/competitions/ml-gen-x-bioreasoning-challenge-track-c/train.csv"
OUTPUT_CSV = "/kaggle/working/rejection_sampled_train_phase_one.csv"

MAX_ROWS = 7705
NUM_TRIALS = 16
BASE_SEED = 42
MAX_TOKENS = 2048
TEMPERATURE = 0.7

trial_columns = [f"trial_{i}" for i in range(1, NUM_TRIALS + 1)]

train_df = pd.read_csv(TRAIN_CSV).head(MAX_ROWS).reset_index(drop=True)
tokenizer = llm.get_tokenizer()


def save_progress(df):
    """Safely write progress without leaving a partially written CSV."""
    temporary_path = OUTPUT_CSV + ".tmp"
    df.to_csv(temporary_path, index=False)
    os.replace(temporary_path, OUTPUT_CSV)


# ---------------------------------------------------------
# Create or resume output dataframe
# ---------------------------------------------------------

if os.path.exists(OUTPUT_CSV):
    saved_df = pd.read_csv(OUTPUT_CSV)

    if len(saved_df) != len(train_df):
        raise ValueError(
            f"Existing output contains {len(saved_df)} rows, "
            f"but current training data contains {len(train_df)} rows."
        )

    # Support resuming if new trial columns were not created previously.
    for column in trial_columns:
        if column not in saved_df.columns:
            saved_df[column] = pd.NA

    if "accepted_trial" not in saved_df.columns:
        saved_df["accepted_trial"] = pd.NA

    completed = saved_df["accepted_trial"].notna().sum()
    print(f"Resuming: {completed}/{len(saved_df)} rows already solved.")

else:
    saved_df = train_df[["pert", "gene", "label"]].copy()
    saved_df.insert(0, "row_id", saved_df.index)

    for column in trial_columns:
        saved_df[column] = pd.NA

    saved_df["accepted_trial"] = pd.NA

    save_progress(saved_df)
    print(f"Created output file with {len(saved_df)} rows.")


# ---------------------------------------------------------
# Run dataset-wide rejection-sampling passes
# ---------------------------------------------------------

for trial_number in range(1, NUM_TRIALS + 1):
    trial_column = f"trial_{trial_number}"

    # A row is unresolved if no previous trial was accepted.
    unresolved_mask = saved_df["accepted_trial"].isna()
    unresolved_indices = saved_df.index[unresolved_mask].tolist()

    if not unresolved_indices:
        print("✅ Every row has been solved.")
        break

    print("=" * 60)
    print(
        f"Starting trial {trial_number}/{NUM_TRIALS}: "
        f"{len(unresolved_indices)} unresolved rows"
    )
    print("Building prompts...")

    prompts = []

    for dataframe_index in unresolved_indices:
        row = saved_df.loc[dataframe_index]

        prompt = build_vllm_prompt(
            pert=row["pert"],
            gene=row["gene"],
            tokenizer=tokenizer,
        )
        prompts.append(prompt)

    # A different seed is necessary; otherwise later passes may reproduce
    # the same sampled output as earlier passes.
    sampling_params = SamplingParams(
        n=1,
        temperature=TEMPERATURE,
        max_tokens=MAX_TOKENS,
        seed=BASE_SEED + trial_number - 1,
    )

    print(
        f"Generating {len(prompts)} outputs in one vLLM request "
        f"using seed {sampling_params.seed}..."
    )

    results = llm.generate(prompts, sampling_params)

    accepted_this_trial = 0

    print("Checking generated answers...")

    for position, (dataframe_index, result) in enumerate(
        zip(unresolved_indices, results),
        start=1,
    ):
        output_text = result.outputs[0].text.strip()
        prediction = parse_label(output_text)
        gold = str(saved_df.at[dataframe_index, "label"]).lower().strip()

        if prediction == gold:
            saved_df.at[dataframe_index, trial_column] = output_text
            saved_df.at[dataframe_index, "accepted_trial"] = trial_number
            accepted_this_trial += 1

        if position % 100 == 0 or position == len(unresolved_indices):
            print(
                f"Checked {position}/{len(unresolved_indices)} | "
                f"Accepted this trial: {accepted_this_trial}"
            )

    save_progress(saved_df)

    total_accepted = saved_df["accepted_trial"].notna().sum()
    remaining = len(saved_df) - total_accepted

    print(
        f"✅ Trial {trial_number} complete | "
        f"New: {accepted_this_trial} | "
        f"Total: {total_accepted} | "
        f"Remaining: {remaining}"
    )
    print(f"Saved progress to: {OUTPUT_CSV}")


print("=" * 60)
print(f"Finished after at most {NUM_TRIALS} trials.")
print(f"Accepted rows: {saved_df['accepted_trial'].notna().sum()}")
print(f"Unresolved rows: {saved_df['accepted_trial'].isna().sum()}")
print(f"Saved to: {OUTPUT_CSV}")
print("=" * 60)
```
